### **Section 1 — NumPy Arrays**
### Concept Recap
Before you start, remember:

| Concept | Key idea |
|---|---|
| **1-D array** | A single row of homogeneous values — `np.array([...])`, `np.arange()`, `np.linspace()`, `np.zeros()`, `np.ones()` |
| **2-D array** | Rows × columns (a table/grid) — created directly, or via `np.zeros((r,c))`, indexed as `arr[row, col]`, sliced as `arr[row, start:end]` |
| **3-D array** | A "stack" of 2-D grids — shape `(depth, rows, cols)`, useful for things like multiple time periods of the same table, or image channels |
| **`.shape`, `.ndim`, `.size`** | Tell you the dimensions, number of dimensions, and total element count |
| **`np.concatenate((a,b), axis=...)`** | Joins arrays along a chosen axis — axis 0 = stack rows, axis 1 = stack columns, axis 2 = stack depth-wise |
| **`.reshape()`** | Re-arranges the same data into a new shape without changing values |
| **Broadcasting** | Numpy automatically "stretches" a smaller array (e.g. a single number, or one row) across a bigger one so you can do `array + 10`, `matrix * row_vector`, etc. without writing loops |
| **`np.where(condition, a, b)`** | Vectorized if/else — picks `a` where the condition is `True`, `b` where it's `False` |

````python
import numpy as np

def make_store_grid(rows):
    return np.array(rows)


def store_totals(grid):
    # TODO: sum across each store's days (one number per store)
    return grid.sum(axis=1)

def daily_totals(grid):
    # TODO: sum across all stores for each day (one number per day)
    return grid.sum(axis=0)

def apply_store_discount(grid, discounts):
    # TODO: broadcast a per-store discount rate across each store's row
    discount_column = discounts.reshape(-1, 1)
    return grid * (1 - discount_column)

def flag_underperformers(store_totals_arr, threshold):
    # TODO: use np.where to label each store as "OK" or "UNDERPERFORMING"
    return np.where(store_totals_arr < threshold, "UNDERPERFORMING", "OK")

def get_store_row(grid, store_index):
    # TODO: slice
    return grid[store_index, :]

def get_day_column(grid, day_index):
    # TODO: slice
    return grid[:, day_index]

# --- Demo / test your functions here ---
sales_grid = make_store_grid([
    [420, 610, 300, 705, 890, 1200, 980],   # Store 0
    [200, 150, 180, 220, 260, 300, 310],    # Store 1
    [900, 850, 940, 1000, 1100, 1250, 1300],# Store 2
    [100, 90, 80, 120, 130, 150, 140],      # Store 3
])
print("Store totals:", store_totals(sales_grid))
print("Daily totals:", daily_totals(sales_grid))
print("With discounts:\n", apply_store_discount(sales_grid, np.array([0.1, 0.0, 0.05, 0.2])))
print("Underperformers:", flag_underperformers(store_totals(sales_grid), threshold=1500))
````

````python
import numpy as np

def stack_weeks(week_grids):
    # TODO: turn a list of 2D (stores x days) arrays into one 3D (weeks x stores x days) array
    expanded_weeks =[
        np.expand_dims(week, axis=0) for week in week_grids
    ]
    return np.concatenate(expanded_weeks, axis=0)

def get_week(cube, week_index):
    # TODO
    return cube[week_index, :, :]  # shape will be 2d (4, 7)

def get_store_across_weeks(cube, store_index):
    # TODO
    return cube[:, store_index, :]  # shape will be 2d (3, 7)

def monthly_total_per_store(cube):
    # TODO
    return cube.sum(axis=(0, 2))  # sum across weeks and days, resulting in shape (4,)

def flatten_all_sales(cube):
    # TODO
    return cube.flatten()  

# --- Demo / test your functions here ---
week1_grid = np.array([
    [420, 610, 300, 705, 890, 1200, 980],
    [200, 150, 180, 220, 260, 300, 310],
    [900, 850, 940, 1000, 1100, 1250, 1300],
    [100, 90, 80, 120, 130, 150, 140],
])
week2_grid = week1_grid * 1.05   # pretend week 2 grew 5%
week3_grid = week1_grid * 0.90   # pretend week 3 dropped 10%

sales_cube = stack_weeks([week1_grid, week2_grid, week3_grid])
print("Cube shape:", sales_cube.shape if sales_cube is not None else None)
# TODO: print cube.ndim, cube.size, and explain each in a comment
print("Cube dimensions (ndim):", sales_cube.ndim)  # Should be 3 (weeks, stores, days)
print("Cube size (total elements):", sales_cube.size)  # Should be 84   

print("Store 2 across all weeks:\n", get_store_across_weeks(sales_cube, 2))
print("Monthly total per store:", monthly_total_per_store(sales_cube))
````

### **Section 2 — Object-Oriented Programming**
#### **Concept Recap**

| Concept | Key idea |
|---|---|
| **Class / Object** | A class is a blueprint; an object is a specific instance built from it |
| **`__init__` and `self`** | `__init__` runs automatically when an object is created and sets up its starting attributes; `self` refers to *this particular* object |
| **Instance methods** | Regular methods that act on `self` (one specific object) |
| **`@classmethod`** | Takes `cls` instead of `self` — often used as an *alternate constructor* that builds and returns a new object |
| **`@staticmethod`** | Takes neither `self` nor `cls` — a plain utility function grouped inside the class for organization |
| **Inheritance** | A child class reuses a parent class's attributes/methods by writing `class Child(Parent):` |
| **`super()`** | Lets a child class call its parent's version of a method (e.g. `super().__init__(...)`) instead of duplicating code |
| **Method overriding** | A child class defining a method with the same name as the parent's replaces it for that child |
| **Polymorphism** | Different classes implement the *same method name* differently — calling code doesn't need to know which exact class it has |

*(Note: this assignment intentionally does not require encapsulation or abstraction — keep attributes as plain public attributes and don't use `@property`, name-mangling, or `abc.ABC`.)*

#### **Shared Base Class (Build this together first, before splitting into Parts A/B/C)**

Implement a `Vehicle` base class with:
- `__init__(self, brand, model, rate_per_day)` storing all three as attributes, plus a running attribute `total_rentals = 0` (how many times this vehicle has been rented).
- An instance method `rental_cost(self, days)` that returns `days * self.rate_per_day` — this is the **default** pricing that subclasses will override.
- An instance method `record_rental(self, days)` that increases `self.total_rentals` by 1 and returns the cost by calling `self.rental_cost(days)` (this way, whichever subclass's `rental_cost` applies gets used automatically — this is polymorphism in action).
- An instance method `display_info(self)` that prints the brand, model, and rate.
- A `@classmethod` `from_listing_string(cls, listing)` that acts as an **alternate constructor**: given a string like `"Toyota,Corolla,45"`, it should split it and return a new `Vehicle` (or subclass) object with those values (as floats/ints where appropriate).
- A `@staticmethod` `is_valid_rental_length(days)` that returns `True` if `days` is a positive whole number, `False` otherwise (call this inside `record_rental` and raise a `ValueError` if it's invalid).

````python
class Vehicle:
    def __init__(self, brand, model, rate_per_day):
        # TODO
        self.brand = brand
        self.model = model
        self.rate_per_day = rate_per_day
        self.total_rentals = 0

    def rental_cost(self, days):
        # TODO: default pricing = days * rate_per_day
        return days * self.rate_per_day

    def record_rental(self, days):
        # TODO: validate days using is_valid_rental_length, then update total_rentals
        # and return self.rental_cost(days)
        if self.is_valid_rental_length(days):
            self.total_rentals += days
            return self.rental_cost(days)
        else:
            raise ValueError("Invalid rental length")

    def display_info(self):
        # TODO
        print(f"Vehicle: {self.brand} {self.model}, Rate per day: ${self.rate_per_day:.2f}, Total rentals: {self.total_rentals}")

    @classmethod
    def from_listing_string(cls, listing):
        # TODO: parse "Brand,Model,Rate" and return cls(brand, model, rate)
        brand, model, rate = [part.strip() for part in listing.split(",")]
        return cls(brand, model, float(rate))

    @staticmethod
    def is_valid_rental_length(days):
        # TODO
        return days > 0 and days <= 30  

# quick sanity check (don't remove)
v = Vehicle.from_listing_string("Toyota,Corolla,45")
v.display_info()
print(v.record_rental(3))

````
