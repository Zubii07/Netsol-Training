# Task 1 — Predict the Output and Explain

## Lab Question

Consider the following Python program:

```python
x = 8
y = 3

result = x + y * 2
print(result)

x = float(x)
print(type(x))

print(x > y and y < 5)
print(x == 8 or y == 10)
print(not(x < y))
```


## Answer

1. The output is:

```text
14
<class 'float'>
True
True
True
```

2. After `float(x)`, the data type of `x` is `float`, and its value is `8.0`.

3. `y * 2` is performed first because multiplication has higher operator precedence than addition. Therefore, `x + y * 2` becomes `8 + (3 * 2)`, which gives `14`.

4. `and` requires both conditions to be true, `or` requires at least one condition to be true, and `not` reverses a Boolean result.

5. The comparison operators used are `>`, `<`, and `==`.

6. The statement `x = float(x)` performs type conversion from `int` to `float`.


# Task 2 — Understand the Control Flow

## Lab Question

1. What is the difference between an `if`, `if-else`, and `if-elif-else` statement?
2. What is indentation in Python, and why is it important?
3. What is the difference between a `for` loop and a `while` loop?
4. What values will be produced by `range(2, 10, 2)`?
5. What is the difference between `break` and `continue`?
6. What does `pass` do?
7. Why does the given age program need a change?


## Answer

1. The `if` statement lets us make decisions based on conditions. If the condition is `true`, it runs a block of code. The `if-else` statement allows us to execute one block of code if a condition is `true` and another block if the condition is `false`. An `if-elif-else` allows multiple conditions to be checked.

2. Indentation means spaces at the beginning of a line. Python uses indentation to identify blocks of code, so it is important for defining program structure.

3. A `for` loop is commonly used to iterate through a sequence or a known range. A `while` loop continues as long as its condition is true.

4. `range(2, 10, 2)` produces `2, 4, 6, 8`. The ending value `10` is not included.

5. `break` completely stops the loop, while `continue` skips the current iteration and moves to the next iteration.

6. `pass` does nothing. It is used as a placeholder where a statement is required but no action is currently needed.

7. `input()` returns a string. Therefore, comparing `age` directly with the integer `18` causes a type error. The input must be converted to an integer using `int()`:

```python
age = int(input("Enter your age: "))

if age >= 18:
    print("Adult")
```


# Practical Task 1 — Simple Student Information & Result

## Lab Question

Write a Python program that asks the user for student name, age, and marks in three subjects. Store the information, convert numeric inputs, calculate total and average, display the result category, check whether the student has passed overall and scored at least 40 in every subject, and display a suitable final message.


````python
# Taking student information
name = input("Enter student name: ")
age = int(input("Enter age: "))

marks1 = float(input("Enter marks in Subject 1: "))
marks2 = float(input("Enter marks in Subject 2: "))
marks3 = float(input("Enter marks in Subject 3: "))

# Calculating total and average
total = marks1 + marks2 + marks3
average = total / 3

# Displaying student information
print("\n Student Result")
print("Name:", name)
print("Age:", age)
print("Total Marks:", total)
print("Average:", average)

# Checking result category
if average >= 80:
    print("Result: Excellent")
elif average >= 60:
    print("Result: Good")
elif average >= 50:
    print("Result: Pass")
else:
    print("Result: Fail")

# Checking overall passing condition
passed_overall = average >= 50
all_subjects_passed = marks1 >= 40 and marks2 >= 40 and marks3 >= 40

if passed_overall and all_subjects_passed:
    print("Final Message: Student has passed successfully.")
else:
    print("Final Message: Student has failed.")
````

# Practical Task 2 — Number Analyzer

## Lab Question

Write a program that asks for a positive integer `N`. Using a `for` loop and `range()`, display numbers from 1 to N, calculate the total sum, even sum, odd sum, count numbers divisible by 3, skip numbers divisible by 5 using `continue`, and stop when the loop reaches 50 using `break`.


````python
N = int(input("Enter a positive integer: "))

total_sum = 0
even_sum = 0
odd_sum = 0
divisible_by_3 = 0

print("\nNumbers:")

for number in range(1, N + 1):
    # Stop when the number reaches 50
    if number == 50:
        break

    # Skip numbers divisible by 5
    if number % 5 == 0:
        continue

    print(number)
    total_sum += number

    if number % 2 == 0:
        even_sum += number
    else:
        odd_sum += number

    if number % 3 == 0:
        divisible_by_3 += 1

print("\n--- Results ---")
print("Total Sum:", total_sum)
print("Even Numbers Sum:", even_sum)
print("Odd Numbers Sum:", odd_sum)
print("Numbers Divisible by 3:", divisible_by_3)
````

# Practical Task 3 — Mini Shopping Bill

## Lab Question

Create a simple shopping-bill program. Ask for the customer name and prices of three items. Calculate the total, apply a 10% discount for totals of 5000 or more, a 5% discount for totals from 3000 to 4999, and no discount otherwise. Display the bill and use a logical condition to check whether the customer qualifies for a discount.


````python
# Taking customer information
customer_name = input("Enter customer name: ")

# Taking prices of three items
price1 = float(input("Enter price of item 1: "))
price2 = float(input("Enter price of item 2: "))
price3 = float(input("Enter price of item 3: "))

# Calculating original total
total = price1 + price2 + price3

# Checking discount
if total >= 5000:
    discount_rate = 10
elif total >= 3000 and total < 5000:
    discount_rate = 5
else:
    discount_rate = 0

# Calculating discount and final amount
discount = total * discount_rate / 100
final_amount = total - discount

# Checking whether customer qualifies for discount
if total >= 3000 and discount_rate > 0:
    qualifies = "Yes"
else:
    qualifies = "No"

# Displaying bill
print("\n SHOPPING BILL")
print("Customer Name:", customer_name)
print("Original Total:", total)
print("Discount:", discount)
print("Discount Rate:", discount_rate, "%")
print("Qualifies for Discount:", qualifies)
print("Final Amount:", final_amount)
````

# Practical Task 4 — Guess the Number

## Lab Question

Set a secret number, for example `7`. Allow the user up to 5 attempts. Display `Correct!` and stop with `break` if the guess is correct. Display `Too high` or `Too low` for incorrect guesses. For zero or negative input, display an invalid-input message and use `continue`. After five unsuccessful attempts, reveal the secret number.


````python
secret_number = 7
attempts = 0

while attempts < 5:
    guess = int(input("Enter your guess: "))

    # Invalid input does not count as an attempt
    if guess <= 0:
        print("Invalid input! Please enter a positive number.")
        continue

    attempts += 1

    if guess == secret_number:
        print("Correct!")
        break
    elif guess > secret_number:
        print("Too high")
    else:
        print("Too low")

    print("Attempts remaining:", 5 - attempts)

if attempts == 5 and guess != secret_number:
    print("\nYou have used all 5 attempts.")
    print("The secret number was:", secret_number)
````

# Practical Task 5 — Simple Menu Program

## Lab Question

Create a menu-driven program that repeatedly displays:

```text
1. Check Number
2. Calculate Square
3. Check Age
4. Exit
```

Option 1 checks positive/negative/zero and even/odd. Option 2 calculates the square using the power operator. Option 3 checks child, teenager, or adult. Option 4 exits using `break`. Invalid menu choices should use `continue`, and `pass` should be used somewhere appropriate.


````python
while True:
    print("\nMENU ")
    print("1. Check Number")
    print("2. Calculate Square")
    print("3. Check Age")
    print("4. Exit")

    choice = input("Enter your choice: ")

    # Option 1 - Check Number
    if choice == "1":
        number = int(input("Enter a number: "))

        if number > 0:
            print("The number is Positive.")
        elif number < 0:
            print("The number is Negative.")
        else:
            print("The number is Zero.")

        if number % 2 == 0:
            print("The number is Even.")
        else:
            print("The number is Odd.")

    # Option 2 - Calculate Square
    elif choice == "2":
        number = float(input("Enter a number: "))
        square = number ** 2
        print("Square:", square)

    # Option 3 - Check Age
    elif choice == "3":
        age = int(input("Enter your age: "))

        if age < 13:
            print("Child")
        elif age <= 19:
            print("Teenager")
        else:
            print("Adult")

    # Option 4 - Exit
    elif choice == "4":
        print("Goodbye!")
        break

    # Invalid choice
    else:
        print("Invalid choice! Please select 1, 2, 3, or 4.")
        continue
````
