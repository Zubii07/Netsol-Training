### 📝 Task 1 — List Basics: Create, Access, Modify

### Background
A Python list is **mutable** and **ordered**. Elements are accessed by zero-based index. Negative indices count from the end (`-1` is the last element). Common methods: `.append()`, `.insert()`, `.remove()`, `.pop()`, `.sort()`, `.reverse()`.

### 🎯 Your Task
Start with:
```python
fruits = ["mango", "banana", "apple", "cherry", "orange"]
```

Perform the following operations **in order**, printing the list after each step:

1. Print the **first** and **last** element using positive and negative indexing.
2. **Append** `"grape"` to the end.
3. **Insert** `"kiwi"` at index 2.
4. **Remove** `"banana"` from the list.
5. **Pop** the element at index 0 and print what was removed.
6. **Sort** the list alphabetically and print it.
7. **Reverse** the sorted list and print it.

> 💡 **Hint:** Remember that `.pop(index)` returns the removed element, so you can capture it: `removed = fruits.pop(0)`.


````python
fruits = ["mango", "banana", "apple", "cherry", "orange"]

# YOUR CODE HERE
# 1. Print first and last elements
print(fruits[0])
print(fruits[-1])
print("List after printing first and last elements:", fruits)


# 2. Append "grape"
fruits.append("grape")
print("List after appending 'grape':", fruits)

# 3. Insert "kiwi" at index 2
fruits.insert(2, "kiwi")
print("List after inserting 'kiwi' at index 2:", fruits)

# 4. Remove "banana"
fruits.remove("banana")
print("List after removing 'banana':", fruits)

# 5. Pop element at index 0, print what was removed
removed = fruits.pop(0)
print(f"Removed: {removed}")
print("List after popping element at index 0:", fruits)

# 6. Sort alphabetically
fruits.sort()
print("List after sorting alphabetically:", fruits)

# 7. Reverse the list
fruits.reverse()
print("List after reversing:", fruits)

print("Final list:", fruits)
````

### 📝 Task 2 — List Comprehensions

### Background
List comprehensions are a concise way to create lists:
```python
[expression for item in iterable if condition]
```
They replace many `for` loops with a single readable line.

### 🎯 Your Task
Use **list comprehensions** (no regular for-loops!) to create and print each of the following:

1. A list of the **squares** of numbers 1 through 15: `[1, 4, 9, 16, ...]`
2. A list of numbers from 1–50 that are **divisible by 3 or 5** (but not both): `[3, 5, 6, 9, 10, ...]`
3. A list of the **lengths** of each word in the sentence `"the quick brown fox jumps over the lazy dog"`
4. A list of **uppercase versions** of only the words that have **5 or more letters**: `['QUICK', 'BROWN', 'JUMPS', 'LAZY']` *(hint: 'over' has 4, skip it)*

> 💡 **Hint for #2:** `(n % 3 == 0) != (n % 5 == 0)` is Python's way to express XOR for two conditions.  
> 💡 **Hint for #3:** `sentence.split()` gives you the word list.


````python
# 1. Squares of 1 to 15
squares = [x**2 for x in range(1, 16)]
print("Squares of 1 to 15:", squares)


# 2. Numbers 1-50 divisible by 3 OR 5 but NOT both
divisible_by_3_or_5 = [x for x in range(1, 51) if (x % 3 == 0) != (x % 5 == 0)]
print("Numbers 1-50 divisible by 3 OR 5 but NOT both:", divisible_by_3_or_5)


# 3. Lengths of each word in the sentence
sentence = "the quick brown fox jumps over the lazy dog"
lengths = [len(word) for word in sentence.split()]
print("Lengths of each word in the sentence:", lengths)


# 4. Uppercase words with 5 or more letters
uppercase_words = [word.upper() for word in sentence.split() if len(word) >= 5]
print("Uppercase words with 5 or more letters:", uppercase_words)
````

### 📝 Task 3 — Tuples: Immutability & Practical Uses

### Background
Tuples look like lists but use `()` instead of `[]`. They are **immutable** — you cannot add, remove, or change elements after creation. This makes them useful for fixed data like coordinates, RGB colours, or records. Tuples support indexing, slicing, and iteration just like lists.

### 🎯 Your Task

**Part A — Basic Tuple Operations:**
```python
student = ("Alice", 22, "Computer Science", 3.8)
```
- Print the student's name (index 0) and GPA (index 3)
- Try to change the name to `"Bob"` — catch the resulting `TypeError` with a `try/except` block and print `"Tuples are immutable!"`
- Unpack the tuple into 4 separate variables: `name, age, major, gpa` and print each

**Part B — List of Tuples:**  
You have coordinates of 5 cities:
```python
cities = [("Karachi", 24.86, 67.01),
          ("Lahore",  31.55, 74.35),
          ("Islamabad", 33.72, 73.06),
          ("Peshawar", 34.01, 71.57),
          ("Quetta",  30.18, 66.99)]
```
- Loop over the list and print each city in the format: `"City: Karachi | Lat: 24.86 | Lon: 67.01"`
- Sort the cities by **latitude** (ascending) and print the sorted list

> 💡 **Hint for sorting:** `sorted(cities, key=lambda c: c[1])` sorts by the second element of each tuple.


````python
# Part A
student = ("Alice", 22, "Computer Science", 3.8)

# Print name and GPA
print(f"Name: {student[0]}, GPA: {student[3]}")


# Try to change name — catch TypeError
try:
    student[0] = "Bob"
except TypeError as e:
    print(f"Error: {e}")


# Unpack into 4 variables and print each
name, age, major, gpa = student
print(f"Name: {name}, Age: {age}, Major: {major}, GPA: {gpa}")


# Part B
cities = [("Karachi",   24.86, 67.01),
          ("Lahore",    31.55, 74.35),
          ("Islamabad", 33.72, 73.06),
          ("Peshawar",  34.01, 71.57),
          ("Quetta",    30.18, 66.99)]

# Loop and print each city's info
for city in cities:
    print(f"City: {city[0]}, Latitude: {city[1]}, Longitude: {city[2]}")


# Sort by latitude and print
sorted_cities = sorted(cities, key=lambda x: x[1])
print("Cities sorted by latitude:")
for city in sorted_cities:
    print(f"City: {city[0]}, Latitude: {city[1]}, Longitude: {city[2]}")
````

### 📝 Task 4 — Dictionary Basics: Create, Access, Update

### Background
A dictionary stores data as **key → value** pairs. Keys must be unique and immutable (strings, numbers, tuples). Values can be anything. Key methods: `.keys()`, `.values()`, `.items()`, `.get()`, `.update()`, `.pop()`.

### 🎯 Your Task
Start with this student record:
```python
student = {
    "name": "Sara",
    "age": 20,
    "courses": ["Math", "Physics", "CS"],
    "grades": {"Math": 88, "Physics": 75, "CS": 92}
}
```

1. Print Sara's **name** and **age**
2. Print her **list of courses**
3. Print her **CS grade** (access the nested dictionary)
4. Add a new key `"year"` with value `2`
5. Update her **age** to `21`
6. Remove the `"Physics"` grade from the nested `"grades"` dictionary using `.pop()`
7. Loop over all key-value pairs with `.items()` and print them line by line
8. Print a list of all **keys** and a list of all **values**

> 💡 **Hint:** Access nested dicts with chained brackets: `student["grades"]["CS"]`  
> 💡 **Hint:** `student.get("email", "Not found")` returns a default value if the key doesn't exist.


````python
student = {
    "name": "Sara",
    "age": 20,
    "courses": ["Math", "Physics", "CS"],
    "grades": {"Math": 88, "Physics": 75, "CS": 92}
}

# 1. Print name and age
print(f"Name: {student['name']}, Age: {student['age']}")

# 2. Print list of courses
print(f"Courses: {student['courses']}")

# 3. Print CS grade (nested access)
grade_cs = student["grades"]["CS"]
print(f"CS Grade: {grade_cs}")

# 4. Add "year": 2
student["year"] = 2
print(f"Year: {student['year']}")

# 5. Update age to 21
student["age"] = 21

# 6. Remove "Physics" from grades using .pop()
student["grades"].pop("Physics", None)

# 7. Loop and print all key-value pairs
for key, value in student.items():
    print(f"{key}: {value}")

# 8. Print list of keys and list of values
print(f"Keys: {list(student.keys())}")
print(f"Values: {list(student.values())}")
````

### 📝 Task 5 — Dictionary Comprehension & Word Frequency Counter

### Background
Just like list comprehensions, you can build dictionaries in one line:
```python
{key_expr: value_expr for item in iterable if condition}
```
Dictionaries are perfect for counting frequencies — a pattern you'll use constantly in data science.

### 🎯 Your Task

**Part A — Dictionary Comprehension:**
1. Create a dictionary mapping each number 1–10 to its **cube**: `{1: 1, 2: 8, 3: 27, ...}`
2. Create a dictionary mapping each number 1–20 to `"even"` or `"odd"`: `{1: 'odd', 2: 'even', ...}`

**Part B — Word Frequency Counter:**  
Count how many times each word appears in the text below, ignoring case. Print the result sorted by **frequency (descending)**.

```python
text = "to be or not to be that is the question to be is to think"
```
Expected top entries: `to: 4, be: 3, is: 2, ...`

> 💡 **Hint for Part B:** Start with `freq = {}`, then loop over `text.lower().split()`.  
> Use `freq[word] = freq.get(word, 0) + 1` to safely increment counts.  
> Sort with `sorted(freq.items(), key=lambda x: x[1], reverse=True)`.


````python
# Part A — Dictionary comprehensions
# 1. Cubes of 1 to 10
cubes = {x: x**3 for x in range(1, 11) }
print("Cubes of 1 to 10:", cubes)


# 2. Even/Odd mapping for 1 to 20
even_odd_mapping = {x: "Even" if x % 2 == 0 else "Odd" for x in range(1, 21)}
print("Even/Odd mapping for 1 to 20:", even_odd_mapping)


# Part B — Word frequency counter
text = "to be or not to be that is the question to be is to think"
frequency = {}
for word in text.lower().split():
    frequency[word] = frequency.get(word, 0) + 1
print("Word frequency counter:", frequency)
````

### 📝 Task 6 — Set Operations

### Background
A **set** is an **unordered collection of unique elements**. Duplicate values are automatically removed. Sets support mathematical operations: union (`|`), intersection (`&`), difference (`-`), symmetric difference (`^`).

| Operation | Symbol | Method | Meaning |
|-----------|--------|--------|---------|
| Union | `\|` | `.union()` | All elements from both sets |
| Intersection | `&` | `.intersection()` | Only elements in **both** |
| Difference | `-` | `.difference()` | Elements in A but **not** B |
| Symmetric Difference | `^` | `.symmetric_difference()` | Elements in either but **not both** |

### 🎯 Your Task
```python
python_students = {"Alice", "Bob", "Carol", "David", "Eve"}
java_students   = {"Bob", "David", "Frank", "Grace", "Alice"}
```

1. Find students enrolled in **both** courses (intersection)
2. Find students enrolled in **at least one** course (union)
3. Find students taking **Python but NOT Java** (difference)
4. Find students taking **exactly one** course (symmetric difference)
5. Check if `{"Bob", "Alice"}` is a **subset** of `python_students`
6. Add `"Henry"` to `python_students` and remove `"Carol"` from it
7. Create a list `[1, 2, 2, 3, 3, 3, 4, 4, 4, 4]` and use a set to print only the **unique values**

> 💡 **Hint:** Use `.issubset()` or the `<=` operator for subset check.


````python
python_students = {"Alice", "Bob", "Carol", "David", "Eve"}
java_students   = {"Bob", "David", "Frank", "Grace", "Alice"}

# 1. Intersection
intersection = python_students.intersection(java_students)
print("Students enrolled in both courses:", intersection)

# 2. Union
union = python_students.union(java_students)
print("Students enrolled in either course:", union)

# 3. Difference (Python but NOT Java)
difference = python_students.difference(java_students)
print("Students enrolled in Python but not Java:", difference)

# 4. Symmetric difference
symmetric_diff = python_students.symmetric_difference(java_students)
print("Students enrolled in either course but not both:", symmetric_diff)

# 5. Subset check: {"Bob", "Alice"} ⊆ python_students
subset_check = {"Bob", "Alice"}.issubset(python_students)
print("Is {'Bob', 'Alice'} a subset of python_students?", subset_check)

# 6. Add "Henry", remove "Carol"
python_students.add("Henry")
python_students.discard("Carol")
print("Updated python_students after adding 'Henry' and removing 'Carol':", python_students)

# 7. Unique values from list
nums = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]
unique_nums = list(set(nums))
print("Unique values from list:", unique_nums)
````
